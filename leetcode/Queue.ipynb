```python
# 17. Letter Combinations of a Phone Number

class Solution:
    def letterCombinations(self, digits):
        if not digits:
            return []

        letter_map = {
            '2': 'abc', '3': 'def', '4': 'ghi',
            '5': 'jkl', '6': 'mno',
            '7': 'pqrs', '8': 'tuv', '9': 'wxyz'
        }

        queue = deque([""])

        for digit in digits:
            for _ in range(len(queue)):
                prev = queue.popleft()
                for char in letter_map[digit]:
                    queue.append(prev + char)

        return list(queue)
```
