```python
# Combination Sum IV
class Solution:
    def combinationSum4(self, nums, target):

        dp = {}

        def backtrack(s):
            if s == 0:
                return 1

            if s < 0:
                return 0

            if s in dp:
                return dp[s]

            ans = 0

            for i in nums:
                ans += backtrack(s - i)

            dp[s] = ans
            return ans

        return backtrack(target)
```
