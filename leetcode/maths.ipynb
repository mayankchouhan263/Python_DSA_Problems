```python
# 3513. Number of Unique XOR Triplets I

class Solution:
    def uniqueXorTriplets(self, nums):
        n = len(nums)

        if n <= 2:
            return n

        return 1 << n.bit_length()
```

```python
# 3514. Number of Unique XOR Triplets II

class Solution:
    def uniqueXorTriplets(self, nums):
        n = len(nums)

        if n == 1:
            return 1

        pair_xors = set()

        for i in range(n):
            for j in range(i + 1, n):
                pair_xors.add(nums[i] ^ nums[j])

        ans = set(nums)  

        for px in pair_xors:
            for x in nums:
                ans.add(px ^ x)

        return len(ans)
```

```python
# 326. Power of Three

class Solution:
    def isPowerOfThree(self, n):
        if n < 1:
            return False

        while n % 3 == 0:
            n //= 3

        return n == 1
```

```python
# 2220. Minimum Bit Flips to Convert Number

class Solution:
    def minBitFlips(self, start, goal):
        x = start ^ goal
        count = 0 
        while x>0:
            if x%2 != 0:
                count += 1
            x = x//2
        return count
```

```python
# 504. Base 7

class Solution:
    def convertToBase7(self, num):
        if num == 0:
            return "0"

        lnum = []
        negative = num < 0
        num = abs(num)

        while num != 0:
            lnum.append(num % 7)
            num //= 7

        lnum.reverse()

        ans = ''.join(map(str, lnum))

        if negative:
            ans = '-' + ans

        return ans
```

```python
# 3702. Longest Subsequence With Non-Zero Bitwise XOR

class Solution:
    def longestSubsequence(self, nums ) :
        xor = 0

        for num in nums:
            xor ^= num

        if xor != 0:
            return len(nums)

        if any(num != 0 for num in nums):
            return len(nums) - 1

        return 0
```
